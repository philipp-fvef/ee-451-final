## Overview

This notebook presents a complete pipeline for detecting and classifying UNO cards in tabletop images. The system solves four connected problems:

1. **Card Segmentation**: Detect individual cards from the full image
2. **Player Assignment**: Assign detected cards to their respective players based on spatial location
3. **Active Player Detection**: Identify which player's turn it is using visual cues
4. **Card Classification**: Classify each detected card using shape-based features

The solution combines traditional image processing (morphological operations, contour detection) with feature matching against a reference database. This hybrid approach is interpretable and efficient, achieving robust performance across varying image conditions.

### Design Rationale

- **Segmentation via morphological operations**: Card borders are distinctive; closing operations connect card regions reliably
- **Color-based feature extraction**: UNO card symbols have strong color properties enabling robust feature matching
- **Fourier descriptors + shape statistics**: Captures both global structure and local contour properties of card symbols
- **Reference-based classification**: Avoids training neural networks on limited data; uses a built reference image database instead

## Segmentation: Card Detection

Card segmentation detects individual cards from the full image using morphological operations on card borders.

### Technical Approach

1. **Border Detection**: Detected via HSV thresholding on the card border color
   - Card borders are gold/yellow; HSV ranges isolate them effectively
   - Thresholded binary mask identifies border pixels

2. **Morphological Closing**: Connects fragmented border regions
   - Closing = Dilation followed by Erosion
   - Fills small holes in the border, creating continuous contours
   - Configurable kernel size balances performance vs. robustness

3. **Contour Extraction**: OpenCV finds card boundaries from the morphological result
   - Contours are filtered by area to remove noise
   - Bounding rectangles approximate card location

4. **Return Coordinates**: Each detected card includes:
   - Pixel coordinates (x, y, width, height)
   - Cropped card image for downstream classification

### Why This Works

- **Gold borders are distinctive**: Even with occlusion or rotation, borders remain visible
- **Morphological operations scale**: Effective across different image resolutions and lighting
- **No learned parameters**: Tunable thresholds but no neural network dependency

## Cropping and Player Assignment

Once cards are segmented, they must be assigned to players. Player positions are fixed (bottom, right, top, left) but card locations vary with hand size and position.

### Technical Approach

1. **Sector Polygons**: Divides the image into four regions, one per player
   - Polygon vertices are determined from image corners and center point
   - Four triangular/trapezoidal sectors radiate from the table center
   - Configurable center point and sector angles

2. **Spatial Assignment**: For each detected card:
   - Compute card centroid from coordinates
   - Determine which sector polygon contains the centroid
   - Assign card to corresponding player

3. **Center Card Detection**: 
   - Cards near the image center (within `center_distance` threshold) are labeled as center cards
   - Separated from player hand cards for special handling

### Why This Works

- **Layout is consistent**: Players always sit in the same relative positions
- **Geometric reasoning is reliable**: Card centroids clearly fall into player regions
- **Configurable thresholds**: Adapt to different table sizes and camera angles

## Active Player Detection

The active player is identified via a visual token (marker) placed next to the current player.

### Technical Approach

1. **Token Localization**: Scans image for a distinctive colored token
   - Token color is typically bright and saturated (configurable in config.json)
   - HSV thresholding isolates token pixels
   - Blob detection finds connected components

2. **Token to Player Mapping**: 
   - Compute token centroid
   - Match to nearest player region using spatial proximity
   - Assign active player label

3. **Fallback Handling**: 
   - If no token is clearly detected, return "EMPTY"
   - Configurable thresholds balance sensitivity vs. false positives

### Why This Works

- **Visual token is unambiguous**: Clearly marks the active player
- **Spatial reasoning**: Token position relative to players determines assignment
- **Robust to occlusion**: Token only needs partial visibility
